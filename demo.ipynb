## 1. Setup

The notebook uses the packaged `src/` folder, so there is no dependency on a personal Desktop path.

```python
from pathlib import Path
import sys, json

PROJECT_ROOT = Path.cwd()
SRC = PROJECT_ROOT / "src"
if str(SRC) not in sys.path:
    sys.path.insert(0, str(SRC))

from cs55_demo import load_config, run_full_pipeline, save_outputs

print("Project root:", PROJECT_ROOT)
```

## 2. Select dataset

The included `config.sample.json` is a tiny smoke test. For the real project demo, change the filename to your local `config.sollevante.json`.

```python
CONFIG_FILE = "config.sample.json"
# Real project example:
# CONFIG_FILE = "config.sollevante.json"

config = load_config(PROJECT_ROOT / CONFIG_FILE)
print(json.dumps({k: str(v) for k, v in config.items()}, indent=2))
```

## 3. Run the full system

The first run generates CLIP embeddings and stores them in `cache/`. Later runs reuse the cache.

```python
cache_path = PROJECT_ROOT / "cache" / f"{config['dataset_name']}_embeddings.pt"

result, bundle = run_full_pipeline(
    config,
    use_cache=True,
    cache_path=cache_path,
)
```

## 4. Final result

```python
print(json.dumps(result, indent=4, ensure_ascii=False))
```

## 5. Save JSON outputs

```python
result_path, bundle_path = save_outputs(
    result,
    bundle,
    PROJECT_ROOT / "outputs",
)

print("Result JSON:", result_path)
print("Evidence-chain JSON:", bundle_path)
```
